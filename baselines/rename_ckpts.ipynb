```python
# ls all files in this directory, and for each pth.tar, open it, and load it with torch.load.

import subprocess
from pathlib import Path
import torch

# list files in the current directory using subprocess
result = subprocess.run(["ls"], capture_output=True, text=True, check=True)
files = result.stdout.split()

# load every .pth.tar file with torch.load
for fname in files:
    if fname.endswith(".pth.tar"):
        print(f"Loading: {fname}")
        obj = torch.load(Path(fname), map_location="cpu")
        print(type(obj))# ls all files in this directory, and for each pth.tar, open it, and load it with torch.load.
```
